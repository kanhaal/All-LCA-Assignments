```python
m1=[[1,2,3],[4,5,6]]
m2=[[4,5,6],[7,8,9]]
res=[[0,0,0],[0,0,0]]
for i in range(len(m1)):
    for j in range(len(m1[0])):
        res[i][j]=m1[i][j]+m2[i][j]
print("Matrix1: ")
for row in m1:
    print(row)
print("\nMatrix2: ")
for row in m2:
    print(row)
print("\nSum of Matrix1 and Matrix2: ")
for row in res:
    print(row)
```

```python
import numpy as np

m1=np.array([[1,2,3],[4,5,6]])
m2=np.array([[4,5,6],[7,8,9]])
res=m1+m2

print("Matrix1:")
print(m1)
print("\nMatrix2:")
print(m2)
print("\nSum of Matrix1 and Matrix2:")
print(res)
```
