```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Student": ["Amit", "Bhavna", "Chetan", "Diya", "Esha", "Farhan"],
    "Maths": [75, 82, 68, 90, 72, 85],
    "Python": [80, 78, 70, 92, 75, 88],
    "Attendance": [85, 90, 78, 95, 80, 92],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai", "Pune"]
}
pd.DataFrame(data).to_csv("sample_data_assignment_07.csv", index=False)
df = pd.read_csv("sample_data_assignment_07.csv")
df
```

```python
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.scatter(df["Maths"], df["Python"])
plt.title("Maths vs Python Marks")
plt.xlabel("Maths")
plt.ylabel("Python")
plt.show()
```

```python
correlation = df[["Maths", "Python", "Attendance"]].corr()
print(correlation)
sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
