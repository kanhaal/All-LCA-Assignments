## Q1

```python
my_dict = {
    "name": "Rahul",
    "rollno": 34,
    "branch": "CSE",
    "age": 18,
    "marks": 85
}

print("Student information:")
print(my_dict)

my_dict["city"] = "Pune"
del my_dict["age"]

print("After adding and deleting:")
print(my_dict)
```

## Q2

```python
list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict = {}

for i in range(len(list1)):
    my_dict[list1[i]] = list2[i]

print(my_dict)
```

## Q4

```python
my_dict = {
    "name": "ABC",
    "rollno": 34,
    "marks": 85,
    "branch": "CSE"
}

keys = []
values = []

for key in my_dict:
    keys.append(key)
    values.append(my_dict[key])

print("Keys:", keys)
print("Values:", values)
```

## Q5

```python
mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

total = 0

for value in mydict.values():
    total = total + value

mean = total / len(mydict)

print("Mean =", mean)
```

## Q7

```python
string = input("Enter a string: ")

frequency = {}

for letter in string:
    if letter in frequency:
        frequency[letter] = frequency[letter] + 1
    else:
        frequency[letter] = 1

print(frequency)
```

## Q8

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Tuple:", result)
```

## Q9

```python
date1 = tuple(map(int, input("Enter first date (dd mm yyyy): ").split()))
date2 = tuple(map(int, input("Enter second date (dd mm yyyy): ").split()))

days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

total1 = date1[2] * 365

for i in range(date1[2]):
    if (i % 4 == 0 and i % 100 != 0) or i % 400 == 0:
        total1 = total1 + 1

for i in range(date1[1] - 1):
    total1 = total1 + days[i]

total1 = total1 + date1[0]

total2 = date2[2] * 365

for i in range(date2[2]):
    if (i % 4 == 0 and i % 100 != 0) or i % 400 == 0:
        total2 = total2 + 1

for i in range(date2[1] - 1):
    total2 = total2 + days[i]

total2 = total2 + date2[0]

difference = total2 - total1

if difference < 0:
    difference = -difference

print("Number of days between two dates:", difference)
```
