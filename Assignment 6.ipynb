```python
df.loc[df["Species"] == "Iris-setosa", ["SepalLengthCm", "SepalWidthCm", "Species"]].head()
```

```python
data = df.copy()
data.loc[0, "SepalLengthCm"] = None
data.loc[1, "Species"] = None
data = pd.concat([data, data.iloc[[2]]], ignore_index=True)
print(data.isnull().sum())
print("Duplicates:", data.duplicated().sum())
```

```python
data["SepalLengthCm"] = data["SepalLengthCm"].fillna(data["SepalLengthCm"].mean())
data["Species"] = data["Species"].fillna(data["Species"].mode()[0])
data = data.drop_duplicates()
print(data.isnull().sum())
print("Duplicates:", data.duplicated().sum())
```

```python
data = data.drop("Id", axis=1)
data.head()
```
